```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.formula.api as smf

# wasn't reading correctly with \t as separator
df = pd.read_csv("salary.txt", sep=r'\s+', engine='python')
```

```python
df["tenure"] = df["year"] - df["startyr"]
df["experience"] = df["year"] - df["yrdeg"]
df["experience2"] = df["experience"] ** 2
```

__2-Sample T-Test for Means:__

```python
male_salaries = df[df["sex"] == "M"]["salary"]
female_salaries = df[df["sex"] == "F"]["salary"]

result = stats.ttest_ind(male_salaries, female_salaries, equal_var=False)
print(f"P-value: {result.pvalue}")
print(f"Average Salary for Males: {np.mean(male_salaries)}")
print(f"Average Salary for Females: {np.mean(female_salaries)}")
print(f"95% Confidence Interval for Difference in Means: ({result.confidence_interval().low}, {result.confidence_interval().high})")
```

A simple 2-sample T-test shows clearly that the mean salaries of men and women in our dataset are not equal.

__T-Tests for Salaries Dependent on Rank:__

```python
for rank in df["rank"].dropna().unique():
    current_df = df[df["rank"] == rank]
    male_salaries = current_df[current_df["sex"] == "M"]["salary"]
    female_salaries = current_df[current_df["sex"] == "F"]["salary"]
    result = stats.ttest_ind(male_salaries, female_salaries, equal_var=False)
    print(f"Rank: {rank}")
    print(f"P-value: {result.pvalue}")
    print(f"Average Salary for Males: {np.mean(male_salaries)}")
    print(f"Average Salary for Females: {np.mean(female_salaries)}")
    print(f"95% Confidence Interval for Difference in Means: ({result.confidence_interval().low}, {result.confidence_interval().high})")
    print("-------")
```

For Assistant and Associate ranks, there does not appear to be significant gender pay differences but there does appear to be a statistically significant difference at the Full rank.
